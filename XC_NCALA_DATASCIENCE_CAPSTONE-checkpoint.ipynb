### Raw data overview: structure, missing values, duplicates and numeric summary

- The dataset has 24,575 rows and 11 columns. Only Year and Age are numeric. kmDriven and AskPrice are stored as text because they contain "km", the rupee sign and commas, so they need converting to numbers before analysis or modelling.
  
- Nine of the eleven columns have missing values. Owner (43%) and model (42%) have the most. kmDriven, PostedDate, Age, Transmission, Brand, FuelType and Year each have between 21% and 25% missing. AdditionInfo and AskPrice are complete, so the target has no gaps. AdditionInfo often contains the brand, model, year and fuel type, so some of the missing values can be recovered from it.
  
- There are 9,587 duplicated records (39% of the data). They need to be dropped, otherwise they inflate the sample and can appear in both the training and test sets.

- Year runs from 1900 to 9999, and the 75th percentile is already 9999, so these are placeholder values. Age runs from 0 to 124, and the maximum matches the year 1900. The medians (Year 2018, Age 7) look reasonable, so the problem is limited to the extreme values. Impossible years need to be set to missing, and Age should be checked against Year afterwards.




### Categorical features: label levels

- Brand has 41 levels. "Unknown" appears alongside NaN, and both mean the brand is missing. "Toyota Land" looks like a truncated label.
  
- model has 403 levels, and the same model is often written in different ways ("Swift-Dzire", "Swift Dzire" and "Dzire"; "Alto-K10" and "Alto K10"; "Wagon-R" and "Wagon R"; "Punch " with a trailing space and "tata-punch"). Some entries are not models at all ("1000", "800", "Others", "Jeep"). The labels need standardising (lowercase, strip spaces, replace hyphens), and rare models may need grouping, otherwise encoding will create hundreds of columns.

  
- Transmission (Manual, Automatic) and Owner (first, second) have clean labels. Their only issue is missing values.

  
- FuelType has "Unknown" and NaN for missing, and "hybrid" and "Hybrid/CNG" are written inconsistently. These need checking before they are merged or kept apart.


- PostedDate is text in a month-year format with 12 levels, and about two thirds of the rows are Nov-24. It needs converting to a date, and it may add little because it varies so little.

```python
# 4. Now we will be moving to a series of steps that help us clean, the dataset and standardise the features the best
# way possible so that they can be fed into the model. 


#Step 1: We remove duplicates and fix the “Unknown” labels

df = Init_used_car_data.copy()

df = df.drop_duplicates().reset_index(drop=True)

df = df.replace("Unknown", np.nan)

print(df.shape)

```

 ->  9,587 exact duplicate rows (39%) were removed, leaving 14,988 records. "Unknown" in Brand and FuelType was converted to NaN so it is counted as missing, the same as the blank entries.

```python

# Step 2: We convert the AskPrice and kmDriven to numbers
df["AskPrice"] = pd.to_numeric(df["AskPrice"].astype(str)
                                            .str.replace("₹", "", regex=False)
                                            .str.replace(",", "", regex=False)
                                            .str.strip(), errors="coerce")

df["kmDriven"] = pd.to_numeric(df["kmDriven"].astype(str)
                                            .str.replace("km", "", regex=False)
                                            .str.replace(",", "", regex=False)
                                            .str.strip(), errors="coerce")

df[["AskPrice", "kmDriven"]].describe()
```

-> Both columns were stored as text (rupee sign, commas and "km"). After conversion, AskPrice is complete, with a median of about ₹570,000. The mean (about ₹997,000) is well above the median and the maximum is ₹42.5 million, so prices are strongly right-skewed. kmDriven has 3,148 missing values and a maximum of 980,000 km, which is unrealistic for a used car and will be handled in the outlier step.

```python
# Step 3: We will be fixing the year and age on records where year and age both exist.

# impossible values become missing
df.loc[(df["Year"] < 1980) | (df["Year"] > 2024), "Year"] = np.nan
df.loc[df["Age"] > 40, "Age"] = np.nan

# rebuild each one from the other
df["Year"] = df["Year"].fillna(2024 - df["Age"])
df["Age"] = df["Age"].fillna(2024 - df["Year"])
print(df[["Year", "Age"]].isnull().sum())
```


 -> What we see here is that Year contained placeholder values (9999) and Age went up to 124, so Year values outside 1980 to 2024 and Age values above 40 were set to missing. In every row where both were present, Year + Age equals 2024, so each column was rebuilt from the other. This left 1,430 rows with both features missing. The Step 4 tries to recover them from the AdditionInfo text.

```python

# Step 4: Here we are recovering the values for Brand, Year and Fueltype from the AdditionInfo feature. 

# Year: a four digit year inside the text
year_text = df["AdditionInfo"].str.extract(r"\b(19[89]\d|20[0-2]\d)\b")[0].astype(float)
df["Year"] = df["Year"].fillna(year_text)
df["Age"] = df["Age"].fillna(2024 - df["Year"])


# Brand: the text usually starts with the brand name
brands = sorted(df["Brand"].dropna().unique(), key=len, reverse=True)

def get_brand(text):
    for b in brands:
        if text.lower().startswith(b.lower()):
            return b
    return np.nan
df["Brand"] = df["Brand"].fillna(df["AdditionInfo"].apply(get_brand))

# FuelType: look for the fuel word in the text
def get_fuel(text):
    t = text.lower()
    if "cng" in t or "hybrid" in t:
        return "Hybrid/CNG"
    if "diesel" in t:
        return "Diesel"
    if "petrol" in t:
        return "Petrol"
    return np.nan

df["FuelType"] = df["FuelType"].fillna(df["AdditionInfo"].apply(get_fuel))
print(df.isnull().sum())
```

 -> AdditionInfo is free text that usually contains the brand, the year and the fuel type (for example "Maruti Suzuki Baleno Alpha CVT, 2019, Petrol"). It was used to fill missing values in three columns. Year missing fell from 1,430 to 82, Brand from 5,215 to 248, and FuelType from 5,237 to 503. As a check, the brand rule matched the existing Brand in about 94% of rows that already had one. Rows where the text had no brand, year or fuel word stay missing and are dealt with in Step 6.

```python

# Step 5: We will be standardising the existing labels on FuelType, Brand, model, PostedDate features or variables.

df["FuelType"] = df["FuelType"].replace({"hybrid": "Hybrid/CNG"})
df["Brand"] = df["Brand"].replace({"Toyota Land": "Toyota"})
df["model"] = (df["model"].str.lower()
                          .str.replace("-", " ", regex=False)
                          .str.strip())
df["PostedDate"] = pd.to_datetime(df["PostedDate"], format="%b-%y")


print(df["FuelType"].value_counts(dropna=False))
print(df["model"].nunique())
```

-> The "hybrid" and "Hybrid/CNG" in FuelType were merged into one label, and the truncated brand "Toyota Land" was changed to "Toyota". Model names were lowercased, stripped of extra spaces and had hyphens replaced, which reduced the model levels from 403 to 388. PostedDate was converted from text to a date. 388 model levels is still a lot, so rare models will need grouping before encoding.

```python

# Step 6: We are here then resolving the remaining missing values and possible outliers 

# drop rows where Year could not be recovered
df = df.dropna(subset=["Year"]).reset_index(drop=True)

# keep missing categories as their own label instead of guessing
for col in ["Brand", "model", "Transmission", "Owner", "FuelType"]:
    df[col] = df[col].fillna("Not stated")

# kmDriven: remove extreme values, then fill gaps with the median
df = df[(df["kmDriven"].isnull()) | (df["kmDriven"] <= 500000)].copy().reset_index(drop=True)
df["kmDriven"] = df["kmDriven"].fillna(df["kmDriven"].median())

print(df.shape)
print(df.isnull().sum())

```


-> With this output step, 82 rows with no recoverable Year were dropped. Missing Brand, model, Transmission, Owner and FuelType were kept as a separate "Not stated" label, since guessing would add false information. 42 rows with kmDriven above 500,000 km were removed as unrealistic, and the remaining missing kmDriven values were filled with the median. High prices were kept because they are genuine luxury cars, and a log transform will be applied to AskPrice for modelling. The cleaned dataset has 14,864 rows. PostedDate still has 3,097 missing values and is mostly one month, so it is likely to be dropped at feature selection.

```python

# Step 7: Proper summary check on the data frame after the various data manipulations for preprocessing 

print("-------------------------Dataset Information---------------")
df.info()

print("---------------------------Numeric Features Summary------------------------------------")
df.describe()
```

-> The cleaned dataset has 14,864 rows and no missing values except PostedDate (3,097). Year ranges from 1986 to 2024 and Age from 0 to 38. Because Age equals 2024 minus Year, the two are perfectly correlated, so only one will be kept for modelling. kmDriven has a median of about 65,000 km and a maximum of 500,000 km after outlier removal. AskPrice has a median of ₹575,000 against a mean of about ₹1.0 million, which confirms a strong right skew. PostedDate is almost entirely November 2024, so it carries little information.

### <u> Focus 2: Modeling and Evaluation </u>

#### - In this part of the model building process, the data will be split into training and test sets, and the categorical features will be encoded. Then we will train a baseline model and an improved model, and evaluate them using cross-validation, RMSE and R². The part ends with tuning the hyperparameters of the best model and comparing its performance before and after tuning.

### Train & Test data Split 

```python

# cell 1 train-test split 

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Train:", X_train.shape, y_train.shape)
print("Test: ", X_test.shape, y_test.shape)
print("Mean LogPrice, train:", round(y_train.mean(), 3), " test:", round(y_test.mean(), 3))
```


-> In this step, the data was split 70/30 into training (10,404 rows) and test (4,460 rows) sets, with a fixed random state for reproducibility. All fitting, including encoding and grouping of rare categories, will use the training set only, and the test set is used once at the end to check performance. The mean LogPrice is almost the same in both sets (13.320 in train and 13.336 in test), so the split is balanced. 

### One-Hot Encoding of the categorical features 

```python
# Cell 2 One-Hot Encoding 

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer([
    ("brand", OneHotEncoder(handle_unknown="infrequent_if_exist", min_frequency=20), ["Brand"]),
    ("model", OneHotEncoder(handle_unknown="infrequent_if_exist", min_frequency=50), ["model"]),
    ("other_cat", OneHotEncoder(handle_unknown="ignore"), ["Transmission", "FuelType", "Owner"]),
    ("num", StandardScaler(), ["Age", "kmDriven"]),
])

X_train_enc = preprocessor.fit_transform(X_train)
X_test_enc = preprocessor.transform(X_test)

print("Before encoding:", X_train.shape)
print("After encoding: ", X_train_enc.shape)
```

#### Baseline model: Linear Regression

```python
#Cell 3 Baseline model

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_validate

lr_pipe = Pipeline([
    ("prep", preprocessor),
    ("model", LinearRegression()),
])

cv_lr = cross_validate(
    lr_pipe, X_train, y_train, cv=5,
    scoring=["neg_root_mean_squared_error", "r2"],
    return_train_score=True,
)

print("Linear Regression, 5-fold cross-validation")
print("Train RMSE:", round(-cv_lr["train_neg_root_mean_squared_error"].mean(), 3))
print("CV RMSE:   ", round(-cv_lr["test_neg_root_mean_squared_error"].mean(), 3))
print("Train R2:  ", round(cv_lr["train_r2"].mean(), 3))
print("CV R2:     ", round(cv_lr["test_r2"].mean(), 3))
```

**Baseline model: Linear Regression**
-> A Linear Regression pipeline (encoding plus model) was scored with 5-fold cross-validation on the training set. The CV RMSE is 0.432 (log units) and the CV R² is 0.778, so the model explains about 78% of the variation in log price. Train and CV scores are almost the same (RMSE 0.428 and R² 0.783 on train), so there is no sign of overfitting. The remaining error is likely from interactions such as brand and age, which a linear model cannot capture.

```python
#Cell 4 Baseline Regression Model Effect 
lr_pipe.fit(X_train, y_train)

lr_model = lr_pipe.named_steps["model"]
coef = pd.DataFrame({
    "feature": lr_pipe.named_steps["prep"].get_feature_names_out(),
    "coefficient": lr_model.coef_,
}).sort_values("coefficient", ascending=False)

print("Intercept:", round(lr_model.intercept_, 3))
print("Number of features:", len(coef))
print("\nHighest coefficients (raise the price)")
print(coef.head(10).to_string(index=False))
print("\nLowest coefficients (lower the price)")
print(coef.tail(10).to_string(index=False))
```

**What drives the price (Linear Regression coefficients)** -> Brand has the largest effect on price. Porsche, Lexus and Land Rover raise the predicted price the most, and Datsun, Chevrolet and Fiat lower it the most. Age also lowers the price, by about 12% per year, which matches the earlier EDA patterns. 

#### Improved model: Random Forest

```python
#Cell 5 Running an alternative random forest algorithm

from sklearn.ensemble import RandomForestRegressor

rf_pipe = Pipeline([
    ("prep", preprocessor),
    ("model", RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)),
])

cv_rf = cross_validate(
    rf_pipe, X_train, y_train, cv=5,
    scoring=["neg_root_mean_squared_error", "r2"],
    return_train_score=True,
)

print("Random Forest, 5-fold cross-validation")
print("Train RMSE:", round(-cv_rf["train_neg_root_mean_squared_error"].mean(), 3))
print("CV RMSE:   ", round(-cv_rf["test_neg_root_mean_squared_error"].mean(), 3))
print("Train R2:  ", round(cv_rf["train_r2"].mean(), 3))
print("CV R2:     ", round(cv_rf["test_r2"].mean(), 3))
```

**Random Forest** -> A Random Forest (200 trees) was scored with the same 5-fold cross-validation. The CV RMSE is 0.416 and the CV R² is 0.795, a small improvement on the baseline (0.432 and 0.778). Training R² is 0.969 against a CV R² of 0.795, so the model is overfitting: it fits the training data much more closely than new data. Tuning the tree depth and leaf size should reduce this.

#### Hyperparameter tuning: Random Forest

```python
#Cell 6 Hyperparameter tuning
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    "model__max_depth": [10, 20, 30, None],
    "model__min_samples_leaf": [1, 3, 5, 10],
    "model__max_features": [0.3, 0.5, 1.0],
}

search = RandomizedSearchCV(
    rf_pipe, param_grid, n_iter=10, cv=5,
    scoring="neg_root_mean_squared_error",
    random_state=42, n_jobs=-1, return_train_score=True,
)
search.fit(X_train, y_train)

best = search.best_index_
print("Best settings:", search.best_params_)
print("Train RMSE:", round(-search.cv_results_["mean_train_score"][best], 3))
print("CV RMSE:   ", round(-search.best_score_, 3))
```

**Hyperparameter tuning: Random Forest** -> RandomizedSearchCV tried 10 combinations of max_depth, min_samples_leaf and max_features with 5-fold cross-validation. The best settings were max_depth=None, min_samples_leaf=1 and max_features=0.5, which lowered the CV RMSE from 0.416 to 0.411. The improvement is small, and the search preferred the least restrictive tree settings, so shrinking the trees did not reduce the gap between train and CV error.

### <u> Focus 3: Finalization and Presentation </u>
### - In this part of the model building process, the final model (the tuned Random Forest) is evaluated once on the test set and compared with the baseline. Then the feature importance and the prediction errors are analysed, and overfitting is discussed. The part ends with the insights, the conclusion and the final presentation.

#### Final evaluation on the test set

```python
#Cell 1 Final evaluation on the test set
from sklearn.metrics import mean_squared_error, r2_score

best_rf = search.best_estimator_

for name, model in [("Linear Regression", lr_pipe), ("Tuned Random Forest", best_rf)]:
    pred = model.predict(X_test)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    r2 = r2_score(y_test, pred)
    pct_error = np.abs(np.exp(pred) - np.exp(y_test)) / np.exp(y_test)
    print(name)
    print("  Test RMSE:", round(rmse, 3))
    print("  Test R2:  ", round(r2, 3))
    print("  Median error:", round(pct_error.median() * 100, 1), "% of the price")
    
```

**Final evaluation on the test set** -> On the test set (4,460 cars), the tuned Random Forest reached an RMSE of 0.397 and an R² of 0.813, against 0.421 and 0.790 for Linear Regression. The median prediction error is 17.1% of the price for the forest and 20.6% for the linear model. Test scores are close to the cross-validation scores, so both models generalise well. The tuned Random Forest is selected as the final model.

**Feature importance (tuned Random Forest)** -> Permutation importance on the test set shows the model relies mainly on Age and Brand, followed by FuelType and Transmission. model and kmDriven add little, and Owner has no effect once the other features are known. This matches the EDA and the linear model coefficients.

#### Error analysis: overfitting check

```python
#Cell 11 Overfitting check
lr_test = np.sqrt(mean_squared_error(y_test, lr_pipe.predict(X_test)))
rf_test = np.sqrt(mean_squared_error(y_test, best_rf.predict(X_test)))

results = pd.DataFrame({
    "model": ["Linear Regression", "Random Forest", "Tuned Random Forest"],
    "Train": [-cv_lr["train_neg_root_mean_squared_error"].mean(),
              -cv_rf["train_neg_root_mean_squared_error"].mean(),
              -search.cv_results_["mean_train_score"][search.best_index_]],
    "CV": [-cv_lr["test_neg_root_mean_squared_error"].mean(),
           -cv_rf["test_neg_root_mean_squared_error"].mean(),
           -search.best_score_],
    "Test": [lr_test, np.nan, rf_test],
}).round(3)
print(results.to_string(index=False))

plot_df = results.melt(id_vars="model", var_name="set", value_name="RMSE")

plt.figure(figsize=(10, 6))
sns.barplot(data=plot_df, x="model", y="RMSE", hue="set")
plt.title("RMSE on train, cross-validation and test sets", fontsize=13)
plt.ylabel("RMSE (log units)")
plt.xlabel("")
plt.legend(title="set", loc="upper left", bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

**Overfitting check** -> Linear Regression has almost the same RMSE on train (0.428), CV (0.432) and test (0.421), so it is not overfitting. Its error is likely limited by being a straight-line model, which cannot capture interactions such as brand and age. The Random Forest fits the training data much more closely (train RMSE 0.163 against CV RMSE 0.416), so it overfits. Tuning did not close this gap (0.162 against 0.411), but the tuned forest still scored best on unseen data (test RMSE 0.397 against 0.421 for the linear model), so it is chosen as the final model.

### Insights and conclusion

The goal was to predict the asking price of a used car from its listing details. After cleaning, 14,864 listings were used. Price depends mostly on Age and Brand, with fuel type and transmission adding some. The tuned Random Forest was the best model, with a test R² of 0.813 and a median error of 17.1% of the price, ahead of Linear Regression (R² of 0.790 and 20.6%). The model is least accurate at the extremes of the price range and for listings priced far from what their features suggest.

Limitations: only 7 features were available, about a fifth of kmDriven values were filled with the median, and many Owner, model and FuelType values were missing. Possible improvements: add features such as kilometres per year, try gradient boosting, fill kmDriven inside the pipeline, and deploy the model as a Streamlit app.
